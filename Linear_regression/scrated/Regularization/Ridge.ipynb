```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python

import numpy as np

np.random.seed(42)
X=np.linspace(-5,5,100).reshape(-1,1)
y = X**2 + np.random.normal(0,2,X.shape)

plt.scatter(X,y)
plt.show()
```

```python

poly = PolynomialFeatures(degree=2)

x_poly = poly.fit_transform(X)
```

```python

model = LinearRegression()
model.fit(x_poly,y)
```

```python

y_ploy = model.predict(x_poly)
```

```python
plt.scatter(X,y)
plt.plot(X,y_ploy)
plt.show()
```

```python
reg = LinearRegression()
```

```python
reg.fit(X,y)
```

```python
y_lin = reg.predict(X)
```

```python

plt.scatter(X,y)
plt.plot(X,y_ploy)
plt.plot(X,y_lin,color = 'red')
plt.show()
```
