```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
data = pd.read_csv("Home_price.csv")
data
```

```python
%matplotlib inline
plt.xlabel("Area")
plt.ylabel("Price")
plt.scatter(data.area,data.price)
```

```python
reg = linear_model.LinearRegression()
reg.fit(data[["area"]] ,data.price)
```

```python
reg.predict([[5020]])  # need to pass 2-D array
```

```python
m=reg.coef_
```

```python
b=reg.intercept_
```

```python
y = ((m*(5020))+b)
y
```

if we have to save the model that we traning on dataset we need to use pickle lib.

```python
import pickle
with open('model_pickle','wb') as f :
    pickle.dump(reg,f)
```

to use that model you save, we need  to use pickle lib

```python
with open("model_pickle",'rb') as f:
    model = pickle.load(f)
```

```python
model.predict([[5020]])
```
